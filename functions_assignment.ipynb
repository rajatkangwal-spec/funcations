1. What is the difference between a function and a method in Python?

A function is a reusable block of code defined independently. A method is a function associated with an object or class and is normally called through that object or class.

Example:
def add(a, b):
    return a + b

class Student:
    def greet(self):
        return "Hello"

add(2, 3) is a function call, while Student().greet() is a method call.

3. What are the different ways to define and call a function in Python?

Functions are commonly defined using def and called by writing the function name followed by parentheses. Python supports positional arguments, keyword arguments, default arguments, *args and **kwargs. A short anonymous function can also be created using lambda.

Example:
def add(a, b=0):

The return statement sends a value back to the code that called the function and ends execution of that function. If a function has no explicit return statement, it returns None.

Example:
def multiply(a, b):

5. What are iterators in Python and how do they differ from iterables?

An iterable is an object that can be traversed, such as a list, tuple, or string. An iterator produces values one at a time and keeps track of its current position using __next__(). The iter() function converts an iterable into an iterator.

Example: a list is iterable, while iter(list) returns an iterator.

6. Explain the concept of generators in Python and how they are defined.

A generator is a special type of iterator that produces values lazily, one at a time. A generator function is defined using yield instead of return.

Example:
def count(n):
    for i in range(n):
        yield i

7. What are the advantages of using generators over regular functions?

Generators are memory-efficient because they do not store all results at once. They calculate and yield each value only when needed. They are especially useful for large datasets, streams, and long or infinite sequences.

Example: a generator can produce numbers one by one instead of creating a large list.

8. What is a lambda function in Python and when is it typically used?

A lambda function is a small anonymous function containing a single expression. It is typically used for short operations, especially with map(), filter(), sorted(), and similar functions.

Example:
square = lambda x: x ** 2

9. Explain the purpose and usage of the map() function in Python.

map() applies a function to every item of an iterable and returns a map iterator. It is useful when each input item needs to be transformed.

Example:
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x*x, numbers))

```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x * x, numbers))
print(squares)
```

10. What is the difference between map(), reduce(), and filter() functions in Python?

map() transforms every item, filter() keeps items that satisfy a condition, and reduce() repeatedly combines items to produce one final result. reduce() is imported from functools.

Example:
map() can double numbers, filter() can keep even numbers, and reduce() can add all numbers together.

```python
from functools import reduce

nums = [1, 2, 3, 4]
print(list(map(lambda x: x * 2, nums)))
print(list(filter(lambda x: x % 2 == 0, nums)))
print(reduce(lambda a, b: a + b, nums))
```

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

```python
def sum_even(numbers):
    return sum(n for n in numbers if n % 2 == 0)

print(sum_even([1, 2, 3, 4, 6]))
```

3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.

```python
def square_list(numbers):
    return [n ** 2 for n in numbers]

print(square_list([1, 2, 3, 4]))
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for n in range(1, 201):
    if is_prime(n):
        print(n, end=" ")
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

```python
class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value

for x in Fibonacci(8):
    print(x, end=" ")
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

print(list(powers_of_two(5)))
```

7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_lines(filename):
    with open(filename, "r") as file:
        for line in file:
            yield line

for line in read_lines("example.txt"):
    print(line, end="")
```

8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

```python
data = [(1, 30), (2, 10), (3, 20)]

result = sorted(data, key=lambda x: x[1])
print(result)
```

```python
celsius = [0, 20, 30, 100]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

10. Create a Python program that uses filter() to remove all the vowels from a given string.

```python
text = "Hello Python"

result = ''.join(filter(lambda ch: ch.lower() not in "aeiou", text))
print(result)
```

11. Imagine an accounting routine used in a book shop. Write a Python program which returns a list with 2-tuples. Each tuple consists of the order number and the product of the price per item and the quantity. The product should be increased by 10,- € if the value of the order is smaller than 100,00 €. Write the program using lambda and map.

```python
orders = [
    [34587, 4, 40.95],
    [98762, 5, 56.80],
    [77226, 3, 32.95],
    [88112, 3, 24.99]
]

def calculate_order(order):
    order_no, quantity, price = order
    total = quantity * price
    if total < 100:
        total += 10
    return (order_no, total)

result = list(map(lambda order: calculate_order(order), orders))
print(result)
```
